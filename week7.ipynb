```python
import nltk
from nltk.tag import hmm
from nltk.corpus import treebank
from nltk.probability import LidstoneProbDist
```

```python
nltk.download('treebank')
nltk.download('universal_tagset')
data = treebank.tagged_sents(tagset='universal')

train_data = [[(word.lower(), tag) for (word, tag) in sent] 
              for sent in data[:3500]]

test_data = [[(word.lower(), tag) for (word, tag) in sent] 
             for sent in data[3500:]]
trainer = hmm.HiddenMarkovModelTrainer()
hmm_tagger = trainer.train_supervised(
    train_data,
    estimator=lambda fd, bins: LidstoneProbDist(fd, 0.1, bins)
)

sentence = "The dog saw the cat".lower().split()

tags = hmm_tagger.tag(sentence)

print("Tagged Sentence:")
print(tags)
```
